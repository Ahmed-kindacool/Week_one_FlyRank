## 2. Target or proxy

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
import re

report = open("../../outputs/model_report.md").read()
rows = re.findall(r"\|\s*(\w+)\s*\|\s*([\d.]+)\s*\|\s*([\d.]+)\s*\|\s*([\d.]+)\s*\|", report)
comparison = pd.DataFrame(rows, columns=["model", "roc_auc", "avg_precision", "precision_at_50"])
for c in ["roc_auc", "avg_precision", "precision_at_50"]:
    comparison[c] = comparison[c].astype(float)
comparison["hits_in_top_50"] = (comparison["precision_at_50"] * 50).round(0).astype(int)
comparison
```
